```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("Libraries imported successfully!")

# Load the dataset directly since it is already in the file system
df = pd.read_csv("/content/Delivery_Logistics (1).csv")
print("Dataset Loaded Successfully")
print("shape:", df.shape)
display(df.head())
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

display(df.head(10))
df.info()

print("\nStatistical Summary:")
display(df.describe())
print("Missing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:",
      df.isnull().sum().sum())

print("\nDuplicate Records:",
      df.duplicated().sum())
text_columns = df.select_dtypes(include='object').columns

for col in text_columns:
    df[col] = df[col].str.strip().str.lower()

# Convert delivery time into numeric hours
for col in ['delivery_time_hours', 'expected_time_hours']:
    extracted = df[col].astype(str).str.extract(r'\.(\d+)$')[0]
    df[col] = pd.to_numeric(extracted, errors='coerce')

# Convert ID into identifier format
df['delivery_id'] = df['delivery_id'].astype(str)
print("Data preparation completed")
display(df.head())
columns = [
    'distance_km',
    'package_weight_kg',
    'delivery_time_hours',
    'expected_time_hours',
    'delivery_rating',
    'delivery_cost'
]

for col in columns:
    print("\nAnalysis of:", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
    print("Average Delivery Time:",
      df['delivery_time_hours'].mean())

print("Average Expected Time:",
      df['expected_time_hours'].mean())

df['delay_hours'] = (
    df['delivery_time_hours'] -
    df['expected_time_hours']
)

print("Average Difference:",
      df['delay_hours'].mean())
print("Average Cost:", df['delivery_cost'].mean())
print("Maximum Cost:", df['delivery_cost'].max())
print("Minimum Cost:", df['delivery_cost'].min())
print("Median Cost:", df['delivery_cost'].median())

cost_by_package = df.groupby(
    'package_type'
)['delivery_cost'].mean()

display(cost_by_package.sort_values(ascending=False))
partner_analysis = df.groupby('delivery_partner').agg(
    total_deliveries=('delivery_id', 'count'),
    average_time=('delivery_time_hours', 'mean'),
    average_cost=('delivery_cost', 'mean'),
    average_rating=('delivery_rating', 'mean')
)

display(partner_analysis)
vehicle_analysis = df.groupby('vehicle_type').agg(
    total_deliveries=('delivery_id', 'count'),
    average_time=('delivery_time_hours', 'mean'),
    average_cost=('delivery_cost', 'mean'),
    average_rating=('delivery_rating', 'mean')
)

display(vehicle_analysis)
df['delayed_flag'] = (
    df['delayed'].astype(str).str.lower() == 'yes'
)

weather_analysis = df.groupby('weather_condition').agg(
    total_deliveries=('delivery_id', 'count'),
    average_time=('delivery_time_hours', 'mean'),
    delay_rate=('delayed_flag', 'mean')
)

weather_analysis['delay_rate'] *= 100

display(weather_analysis)
region_analysis = df.groupby('region').agg(
    total_deliveries=('delivery_id', 'count'),
    average_time=('delivery_time_hours', 'mean'),
    average_cost=('delivery_cost', 'mean'),
    average_rating=('delivery_rating', 'mean')
)

display(region_analysis)
numeric_columns = [
    'distance_km',
    'package_weight_kg',
    'delivery_time_hours',
    'expected_time_hours',
    'delivery_rating',
    'delivery_cost'
]

correlation_matrix = df[numeric_columns].corr()

display(correlation_matrix)
plt.figure(figsize=(9,5))

plt.hist(df['delivery_time_hours'].dropna(),
         bins=15, edgecolor='black')

plt.title('Delivery Time Distribution')
plt.xlabel('Delivery Time (Hours)')
plt.ylabel('Number of Deliveries')

plt.show()
plt.figure(figsize=(9,5))

plt.hist(df['delivery_cost'].dropna(),
         bins=20, edgecolor='black')

plt.title('Delivery Cost Distribution')
plt.xlabel('Delivery Cost')
plt.ylabel('Number of Deliveries')

plt.show()
plt.figure(figsize=(11,6))

sns.barplot(
    data=df,
    x='delivery_partner',
    y='delivery_time_hours',
    errorbar=None
)

plt.title('Average Delivery Time by Partner')
plt.xlabel('Delivery Partner')
plt.ylabel('Average Delivery Time')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='vehicle_type',
    y='delivery_time_hours',
    errorbar=None
)

plt.title('Vehicle Performance Analysis')
plt.xlabel('Vehicle Type')
plt.ylabel('Average Delivery Time')

plt.xticks(rotation=30)
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='vehicle_type',
    y='delivery_time_hours',
    errorbar=None
)

plt.title('Vehicle Performance Analysis')
plt.xlabel('Vehicle Type')
plt.ylabel('Average Delivery Time')

plt.xticks(rotation=30)
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='vehicle_type',
    y='delivery_time_hours',
    errorbar=None
)

plt.title('Vehicle Performance Analysis')
plt.xlabel('Vehicle Type')
plt.ylabel('Average Delivery Time')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='weather_condition',
    y='delayed_flag',
    errorbar=None
)

plt.title('Delivery Delay Rate by Weather')
plt.xlabel('Weather Condition')
plt.ylabel('Delay Rate')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='weather_condition',
    y='delayed_flag',
    errorbar=None
)

plt.title('Delivery Delay Rate by Weather')
plt.xlabel('Weather Condition')
plt.ylabel('Delay Rate')

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
plt.figure(figsize=(9,5))

sns.barplot(
    data=df,
    x='region',
    y='delivery_cost',
    errorbar=None
)

plt.title('Average Delivery Cost by Region')
plt.xlabel('Region')
plt.ylabel('Average Delivery Cost')
plt.show()
plt.figure(figsize=(9,5))

sns.scatterplot(
    data=df,
    x='distance_km',
    y='delivery_cost',
    alpha=0.5
)

plt.title('Distance vs Delivery Cost')
plt.xlabel('Distance (KM)')
plt.ylabel('Delivery Cost')

plt.show()
plt.figure(figsize=(10,7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Logistics Correlation Heatmap')
plt.tight_layout()
plt.show()
print(df['delivery_status'].value_counts())

plt.figure(figsize=(8,5))

sns.countplot(data=df, x='delivery_status')

plt.title('Delivery Status Distribution')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Deliveries')

plt.show()
outlier_columns = [
    'distance_km',
    'package_weight_kg',
    'delivery_time_hours',
    'delivery_cost'
]

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)
    ]

    print("\nColumn:", col)
    print("Outlier Count:", len(outliers))
    status_comparison = pd.crosstab(
    df['delayed'],
    df['delivery_status']
)

display(status_comparison)
print("========== LOGISTICS PERFORMANCE ==========")

print("Total Deliveries:", len(df))

print("Average Delivery Time:",
      round(df['delivery_time_hours'].mean(), 2))

print("Average Delivery Cost:",
      round(df['delivery_cost'].mean(), 2))

print("Average Customer Rating:",
      round(df['delivery_rating'].mean(), 2))

print("Delay Percentage:",round(df['delayed_flag'].mean() * 100, 2), "%")

print("===========================================")
df.to_csv('Cleaned_Delivery_Logistics.csv', index=False)

print("Cleaned dataset exported successfully!")
files.download('Cleaned_Delivery_Logistics(1).csv')
```
